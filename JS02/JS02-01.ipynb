# Step 3 - Visual Inspection

```python
num_cols = ['Age', 'Fare']
for col in num_cols:
  plt.figure()
  plt.hist(df[col], bins=30)
  plt.title(f"distribusi {col}")
  plt.xlabel(col); plt.ylabel("frequency")
  plt.show()
```

```python
# OUTLIER INSPECTION

plt.figure()
plt.boxplot(df['Fare'].dropna(), vert=True)
plt.title("boxplot Fare")
plt.ylabel("Fare")
plt.show()
```

```python
# Inspection of the Composition of Qualitative Variables

cat_cols = ['Survived','Pclass','Sex','Embarked']

fig, axes = plt.subplots(2, 2, figsize=(10, 8))
axes = axes.flatten()

for i, col in enumerate(cat_cols):
    df[col].value_counts(dropna=False).plot(kind='bar', ax=axes[i])
    axes[i].set_title(f'Distribusi {col}')
    axes[i].set_xlabel(col)
    axes[i].set_ylabel('Count')

plt.tight_layout()
plt.show()
```

```python
# HEATMAP
import seaborn as sns

num_only = df.select_dtypes(include=[np.number])
corr = num_only.corr(numeric_only=True)

plt.figure(figsize=(8, 6))
sns.heatmap(corr, cmap='coolwarm', vmin=-1, vmax=1, annot=True, fmt=".2f") #seaborn
plt.title('Correlation Heatmap (numeric)')
plt.tight_layout()
plt.show()
```

```python
# SCATTER PLOT
plt.figure()
surv_map = {0:'No', 1:'Yes'}
plt.scatter(df['Age'], df['Fare'], alpha=0.7)
plt.xlabel('Age'); plt.ylabel('Fare')
plt.title('Scatter Age vs Fare')
plt.show()
```
