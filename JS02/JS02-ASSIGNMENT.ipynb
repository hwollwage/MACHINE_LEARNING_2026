```python
df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/datasets/wbc.csv")
df.head()
```

```python
# 1. Differentiate between variables that are usable and variables that are not usable.

df = df.drop(columns=['id'])
df.head()
print(df.columns.tolist()) # id sekarang gak ada
```

```python
# 2. Perform encoding on the "diagnosis" column.

le = LabelEncoder()

df['diagnosis'] = le.fit_transform(df['diagnosis'])
df['diagnosis'].head()

# print(le.classes_) BENIGN = 0, MALIGNANT = 1
```

```python
# 3. Perform standardization on all columns that possess numerical values

std = StandardScaler()

num_cols = df.select_dtypes(include=['number']).columns

num_cols = num_cols.drop('diagnosis')

df[num_cols] = std.fit_transform(df[num_cols])

df.head(25)
```
