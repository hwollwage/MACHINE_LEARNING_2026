```python
import pandas as pd
import numpy as np
from google.colab import drive
drive.mount('/content/drive')
df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/datasets/voice.csv')
print(df.shape)
print(df.isnull().sum().sum(), 'missing values')
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder

X = df.iloc[:, :-1]
y = LabelEncoder().fit_transform(df.iloc[:, -1])
print(X.shape, np.bincount(y))
```

```python
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

splits = {'70:30': 0.30, '80:20': 0.20}
kernels = ['linear', 'poly', 'rbf']

rows = []
for split_name, test_size in splits.items():
    X_tr, X_te, y_tr, y_te = train_test_split(
        X, y, test_size=test_size, random_state=42, stratify=y)

    for k in kernels:
        clf = make_pipeline(StandardScaler(), SVC(kernel=k))
        clf.fit(X_tr, y_tr)
        rows.append({
            'Split': split_name,
            'Kernel': k,
            'Acc Train': accuracy_score(y_tr, clf.predict(X_tr)),
            'Acc Test': accuracy_score(y_te, clf.predict(X_te)),
        })

hasil = pd.DataFrame(rows)
hasil
```

```python
tabel = hasil.pivot(index='Kernel', columns='Split', values='Acc Test').round(4)
tabel
```
